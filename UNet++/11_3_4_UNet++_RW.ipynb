```python
import os, cv2, torch, numpy as np
from torch.utils.data import Dataset

class RailDataset(Dataset):
    def __init__(self, defect_path, transform=None, skip_empty_masks=True):
        self.defect_path = defect_path
        self.image_dir = os.path.join(defect_path, "Images")
        self.mask_dir  = os.path.join(defect_path, "masks")
        self.transform = transform

        self.images = []

        for f in sorted(os.listdir(self.image_dir)):
            if not f.lower().endswith(('.png', '.jpg', '.jpeg')):
                continue

            img_path  = os.path.join(self.image_dir, f)
            name, _   = os.path.splitext(f)
            mask_path = os.path.join(self.mask_dir, name + ".png")

            if not os.path.exists(mask_path):
                continue

            if skip_empty_masks:
                mask = cv2.imread(mask_path, 0)
                if mask is None or np.sum(mask) == 0:
                    continue

            self.images.append(img_path)

        print(f"Loaded {len(self.images)} non-empty mask samples")

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_path = self.images[idx]
        img_name = os.path.basename(img_path)
        name, _ = os.path.splitext(img_name)

        mask_path = os.path.join(self.mask_dir, name + ".png")

        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        mask = cv2.imread(mask_path, 0)
        if mask is None:
            raise FileNotFoundError(f"Mask missing for {img_name}")

        mask = (mask > 0).astype("float32")

        if self.transform:
            augmented = self.transform(image=image, mask=mask)
            image = augmented["image"]
            mask = augmented["mask"]
        else:
            image = torch.from_numpy(image).permute(2,0,1).float()

        mask = torch.tensor(mask).unsqueeze(0)

        return image, mask, img_name
```

```python
def show_samples(dataset, n=3):
    plt.figure(figsize=(8, 3*n))

    for i in range(n):
        img, mask, _ = dataset[i]   # unpack image, mask, filename

        plt.subplot(n, 2, 2*i+1)
        plt.imshow(img.permute(1,2,0).numpy() / 255.0)  # ✅ FIX
        plt.title("Image")
        plt.axis("off")

        plt.subplot(n, 2, 2*i+2)
        plt.imshow(mask.squeeze(), cmap="gray")
        plt.title("Mask")
        plt.axis("off")

    plt.show()


```

STEP 5️⃣ – MODEL, LOSS, METRICS ✅ (POINT 3)

```python
model = smp.UnetPlusPlus(
    encoder_name="efficientnet-b0",
    encoder_weights="imagenet",
    in_channels=3,
    classes=1
).to(DEVICE)

```

```python
from scipy.ndimage import distance_transform_edt

def boundary_loss(pred, gt):
    """
    Normalized boundary loss (stable).
    """
    gt_np = gt.detach().cpu().numpy()

    # Distance transform
    dist = distance_transform_edt(1 - gt_np)

    # Normalize distance map to [0,1]
    dist = dist / (dist.max() + 1e-8)

    dist = torch.tensor(dist, device=pred.device).float()

    pred = torch.sigmoid(pred)

    return (pred * dist).mean()


```

```python
# =========================
# OPTION 2: FOCAL TVERSKY LOSS
# =========================

bce = nn.BCEWithLogitsLoss()

tversky = smp.losses.TverskyLoss(
    mode="binary",
    alpha=0.7,
    beta=0.3
)

def loss_fn(pred, gt):
    return (
        0.50 * bce(pred, gt) +
        0.40 * tversky(pred, gt) +
        0.10 * boundary_loss(pred, gt)
    )


```

```python
def dice_score(pred, gt, eps=1e-7):
    pred = (torch.sigmoid(pred) > 0.5).float()
    inter = (pred * gt).sum()
    union = pred.sum() + gt.sum()
    return (2*inter + eps) / (union + eps)
```

Hausdorff distance

```python
from scipy.spatial.distance import directed_hausdorff
import numpy as np
```

```python
def hausdorff_distance(pred_mask, gt_mask):
    """
    Computes symmetric Hausdorff Distance (in pixels)
    pred_mask, gt_mask: binary numpy arrays (H,W)
    """
    pred_pts = np.argwhere(pred_mask > 0)
    gt_pts   = np.argwhere(gt_mask > 0)

    # If either mask is empty, HD is undefined → return large value
    if len(pred_pts) == 0 or len(gt_pts) == 0:
        return np.nan

    hd1 = directed_hausdorff(pred_pts, gt_pts)[0]
    hd2 = directed_hausdorff(gt_pts, pred_pts)[0]

    return max(hd1, hd2)
```

6.1 METRIC FUNCTIONS (ADD ONCE)

```python
def iou_score(pred, gt, eps=1e-7):
    pred = (torch.sigmoid(pred) > 0.5).float()
    inter = (pred * gt).sum()
    union = pred.sum() + gt.sum() - inter
    return (inter + eps) / (union + eps)

def pixel_accuracy(pred, gt):
    pred = (torch.sigmoid(pred) > 0.5).float()
    correct = (pred == gt).float().sum()
    total = torch.numel(pred)
    return correct / total

def precision_score(pred, gt, eps=1e-7):
    pred = (torch.sigmoid(pred) > 0.5).float()
    tp = (pred * gt).sum()
    fp = (pred * (1 - gt)).sum()
    return (tp + eps) / (tp + fp + eps)

def recall_score(pred, gt, eps=1e-7):
    pred = (torch.sigmoid(pred) > 0.5).float()
    tp = (pred * gt).sum()
    fn = ((1 - pred) * gt).sum()
    return (tp + eps) / (tp + fn + eps)
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=LR)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="max",
    factor=0.5,
    patience=3,
    #verbose=True
)


start_time = time.time()

```

```python
history = {
    # Loss
    "train_loss": [],
    "val_loss": [],

    # Dice & IoU
    "train_dice": [],
    "val_dice": [],
    "train_iou": [],
    "val_iou": [],

    # Accuracy / Precision / Recall
    "train_acc": [],
    "val_acc": [],
    "train_prec": [],
    "val_prec": [],
    "train_rec": [],
    "val_rec": [],

    # Hausdorff
    "train_hd": [],
    "val_hd": []
}
```

```python
best_val_dice = 0.0
epoch_times = []
total_training_start = time.time()

for epoch in range(EPOCHS):
    epoch_start = time.time()

    # ================= TRAIN =================
    model.train()
    tl = td = ti = tacc = tprec = trec = 0.0

    train_hd = []

    for imgs, masks, _ in train_loader:   # ✅ FIXED
        imgs, masks = imgs.to(DEVICE), masks.to(DEVICE)

        preds = model(imgs)
        loss = loss_fn(preds, masks)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        tl += loss.item()
        td += dice_score(preds, masks).item()
        ti += iou_score(preds, masks).item()
        tacc  += pixel_accuracy(preds, masks).item()
        tprec += precision_score(preds, masks).item()
        trec  += recall_score(preds, masks).item()


        # Hausdorff (TRAIN)
        preds_bin = (torch.sigmoid(preds) > 0.5).float()
        for i in range(preds_bin.size(0)):
            pred_np = preds_bin[i, 0].cpu().numpy()
            gt_np   = masks[i, 0].cpu().numpy()
            hd = hausdorff_distance(pred_np, gt_np)
            if not np.isnan(hd):
                train_hd.append(hd)

    tl /= len(train_loader)
    td /= len(train_loader)
    ti /= len(train_loader)
    tacc  /= len(train_loader)
    tprec /= len(train_loader)
    trec  /= len(train_loader)

    mean_train_hd = np.mean(train_hd) if len(train_hd) > 0 else np.nan
    history["train_hd"].append(mean_train_hd)

    # ================= VALIDATION =================
    model.eval()
    vl = vd = vi = vacc = vprec = vrec = 0.0
    val_hd = []

    with torch.no_grad():
        for imgs, masks, _ in val_loader:   # ✅ FIXED
            imgs, masks = imgs.to(DEVICE), masks.to(DEVICE)
            preds = model(imgs)

            vl += loss_fn(preds, masks).item()
            vd += dice_score(preds, masks).item()
            vi += iou_score(preds, masks).item()
            vacc += pixel_accuracy(preds, masks).item()
            vprec += precision_score(preds, masks).item()
            vrec += recall_score(preds, masks).item()

            # Hausdorff (VAL)
            preds_bin = (torch.sigmoid(preds) > 0.5).float()
            for i in range(preds_bin.size(0)):
                pred_np = preds_bin[i, 0].cpu().numpy()
                gt_np   = masks[i, 0].cpu().numpy()
                hd = hausdorff_distance(pred_np, gt_np)
                if not np.isnan(hd):
                    val_hd.append(hd)

    vl /= len(val_loader)
    vd /= len(val_loader)
    vi /= len(val_loader)
    vacc /= len(val_loader)
    vprec /= len(val_loader)
    vrec /= len(val_loader)
    mean_val_hd = np.mean(val_hd) if len(val_hd) > 0 else np.nan
    history["val_hd"].append(mean_val_hd)

    scheduler.step(vd)

    # ================= SAVE HISTORY =================
    history["train_loss"].append(tl)
    history["val_loss"].append(vl)
    history["train_dice"].append(td)
    history["val_dice"].append(vd)
    history["train_iou"].append(ti)
    history["val_iou"].append(vi)
    history["train_acc"].append(tacc)
    history["val_acc"].append(vacc)
    history["train_prec"].append(tprec)
    history["val_prec"].append(vprec)
    history["train_rec"].append(trec)
    history["val_rec"].append(vrec)





    # ================= SAVE BEST MODEL =================
    if vd > best_val_dice:
        best_val_dice = vd
        checkpoint = {
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_dice": best_val_dice,
            "architecture": "Unet++",
            "encoder": "efficientnet-b0",
            "defect": "Rail Wear",
            "image_size": 512
        }
        torch.save(checkpoint, "best_unetpp_effb0_Rail Wear.pth")

    # ================= RUNTIME =================
    epoch_time = time.time() - epoch_start
    epoch_times.append(epoch_time)

    # ================= PRINT METRICS (COMPLETE) =================
    print(
    f"Epoch [{epoch+1}/{EPOCHS}] | Time {epoch_time:.1f}s\n"
    f"  Train → Loss={tl:.4f}, Dice={td:.4f}, IoU={ti:.4f}, "
    f"Acc={tacc:.4f}, Prec={tprec:.4f}, Rec={trec:.4f}, HD={mean_train_hd:.2f}px\n"
    f"  Val   → Loss={vl:.4f}, Dice={vd:.4f}, IoU={vi:.4f}, "
    f"Acc={vacc:.4f}, Prec={vprec:.4f}, Rec={vrec:.4f}, HD={mean_val_hd:.2f}px\n"
    )


# ================= TOTAL TIME =================
total_training_time = time.time() - total_training_start
print("Total training time:", total_training_time)
print("Avg epoch time:", sum(epoch_times) / len(epoch_times))


```

```python
import numpy as np
import torch

def find_best_iou_threshold(model, val_loader):
    model.eval()
    thresholds = np.arange(0.4, 0.8, 0.05)

    best_iou = 0.0
    best_t = 0.5

    with torch.no_grad():
        for t in thresholds:
            iou_sum = 0.0

            for imgs, masks, _ in val_loader:   # ✅ FIX HERE
                imgs, masks = imgs.to(DEVICE), masks.to(DEVICE)
                probs = torch.sigmoid(model(imgs))
                preds = (probs > t).float()
                iou_sum += iou_score(preds, masks).item()

            mean_iou = iou_sum / len(val_loader)

            print(f"Threshold {t:.2f} → IoU {mean_iou:.4f}")

            if mean_iou > best_iou:
                best_iou = mean_iou
                best_t = t

    print(f"\n✔ Best threshold = {best_t:.2f}, IoU = {best_iou:.4f}")
    return best_t, best_iou
```

```python
import time
import numpy as np
import torch
import torch.nn as nn
from scipy.spatial.distance import directed_hausdorff

```

```python
def hausdorff_distance_95(pred_mask, gt_mask):
    pred_pts = np.argwhere(pred_mask > 0)
    gt_pts   = np.argwhere(gt_mask > 0)

    if len(pred_pts) == 0 or len(gt_pts) == 0:
        return np.nan

    d1 = directed_hausdorff(pred_pts, gt_pts)[0]
    d2 = directed_hausdorff(gt_pts, pred_pts)[0]

    return np.percentile([d1, d2], 95)
```

Severity Visualization and Recommended Action

```python
# ============================================================
# RAIL WEAR SEVERITY CALCULATION (VISION-BASED)
# As per MM Agarwal / Indian Railways (Section 3.9.1)
# ============================================================

import numpy as np

# ------------------------------------------------------------
# CALIBRATION (adjust if camera setup changes)
# ------------------------------------------------------------
PIXEL_TO_MM = 0.25   # 1 pixel = 0.25 mm (example)

# ------------------------------------------------------------
# SEVERITY THRESHOLDS (engineering-aligned)
# ------------------------------------------------------------
LOW_WEAR_MM       = 2.0     # Minor wear
MODERATE_WEAR_MM  = 4.0     # Significant wear
# > 4 mm → Severe wear

# ------------------------------------------------------------
# STEP 1: ESTIMATE MAXIMUM WEAR DEPTH (mm)
# ------------------------------------------------------------
def estimate_rail_wear_depth_mm(mask, pixel_to_mm=PIXEL_TO_MM):
    """
    mask : binary rail wear mask [H,W] or [1,H,W]
    """

    if mask.ndim == 3:
        mask = mask[0]

    ys, xs = np.where(mask > 0)

    if len(ys) == 0:
        return 0.0

    wear_depth_pixels = ys.max() - ys.min()
    wear_depth_mm = wear_depth_pixels * pixel_to_mm

    return wear_depth_mm


# ------------------------------------------------------------
# STEP 2: ESTIMATE WEAR AREA RATIO
# ------------------------------------------------------------
def estimate_wear_area_ratio(mask):
    """
    Returns fraction of rail head affected
    """

    if mask.ndim == 3:
        mask = mask[0]

    total_pixels = mask.size
    wear_pixels = np.sum(mask > 0)

    return wear_pixels / total_pixels


# ------------------------------------------------------------
# STEP 3: RAIL WEAR SEVERITY SCORE
# ------------------------------------------------------------
def compute_rail_wear_severity_score(depth_mm, area_ratio):
    """
    Weighted score combining depth and spread of wear
    """

    # Normalize depth score
    depth_score = min(depth_mm / MODERATE_WEAR_MM, 1.0)

    # Area score (spread of wear)
    area_score = min(area_ratio / 0.15, 1.0)

    # Combined severity score
    severity_score = 0.7 * depth_score + 0.3 * area_score

    return round(severity_score, 3)


# ------------------------------------------------------------
# STEP 4: SEVERITY CLASSIFICATION
# ------------------------------------------------------------
def classify_rail_wear_severity(depth_mm):
    if depth_mm <= LOW_WEAR_MM:
        return "Low"
    elif depth_mm <= MODERATE_WEAR_MM:
        return "Moderate"
    else:
        return "Severe"


# ------------------------------------------------------------
# STEP 5: ACTION RECOMMENDATION (MM Agarwal aligned)
# ------------------------------------------------------------
def recommend_rail_wear_action(severity):
    if severity == "Low":
        return "Monitor during routine inspection"
    elif severity == "Moderate":
        return "Rail grinding / profile correction"
    else:
        return "Rail renewal / replacement"


# ------------------------------------------------------------
# STEP 6: END-TO-END RAIL WEAR SEVERITY FUNCTION
# ------------------------------------------------------------
def rail_wear_severity_analysis(pred_mask):
    """
    pred_mask : binary predicted rail wear mask
    """

    depth_mm = estimate_rail_wear_depth_mm(pred_mask)
    area_ratio = estimate_wear_area_ratio(pred_mask)

    severity_score = compute_rail_wear_severity_score(depth_mm, area_ratio)
    severity_label = classify_rail_wear_severity(depth_mm)
    action = recommend_rail_wear_action(severity_label)

    return {
        "max_wear_depth_mm": round(depth_mm, 2),
        "wear_area_ratio": round(area_ratio, 4),
        "severity_score": severity_score,
        "severity": severity_label,
        "recommended_action": action
    }


# ------------------------------------------------------------
# EXAMPLE USAGE
# ------------------------------------------------------------
"""
result = rail_wear_severity_analysis(pred_mask)

print("===== RAIL WEAR SEVERITY REPORT =====")
print("Max Wear Depth (mm):", result["max_wear_depth_mm"])
print("Wear Area Ratio    :", result["wear_area_ratio"])
print("Severity Score     :", result["severity_score"])
print("Severity Level     :", result["severity"])
print("Action             :", result["recommended_action"])
"""
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Helper: Denormalize image for display
# ------------------------------------------------------------
def denormalize_image(img, mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)):
    img = img.copy()
    for i in range(3):
        img[:, :, i] = img[:, :, i] * std[i] + mean[i]
    img = np.clip(img, 0, 1)
    return (img * 255).astype(np.uint8)


# ------------------------------------------------------------
# Visualization + Action Recommendation for Rail Wear
# ------------------------------------------------------------
def visualize_rail_wear_5_samples(
    model,
    dataloader,
    device,
    threshold=0.25
):
    model.eval()

    images = []
    gt_masks = []
    pred_masks = []
    infos = []
    image_names = []

    # ---------------- Collect 5 samples ----------------
    with torch.no_grad():
        for imgs, gt_msks, filenames in dataloader:
            imgs = imgs.to(device)

            logits = model(imgs)
            preds = (torch.sigmoid(logits) > threshold).float()

            for i in range(imgs.size(0)):
                img = imgs[i].cpu().permute(1, 2, 0).numpy()
                img = denormalize_image(img)

                gt_mask = gt_msks[i][0].cpu().numpy()
                pred_mask = preds[i][0].cpu().numpy()

                # Skip empty GT masks (IMPORTANT for Rail Wear)
                if gt_mask.sum() == 0:
                    continue

                # ---- Rail Wear severity (already implemented by you) ----
                result = rail_wear_severity_analysis(pred_mask)

                images.append(img)
                gt_masks.append(gt_mask)
                pred_masks.append(pred_mask)
                infos.append(result)
                image_names.append(filenames[i])

                if len(images) == 5:
                    break
            if len(images) == 5:
                break

    # ---------------- Plot (5 × 3 grid) ----------------
    plt.figure(figsize=(18, 20))

    for i in range(len(images)):
        # Original image
        plt.subplot(5, 3, 3*i + 1)
        plt.imshow(images[i])
        plt.title("Original Image")
        plt.axis("off")

        # Ground truth
        plt.subplot(5, 3, 3*i + 2)
        plt.imshow(gt_masks[i], cmap="gray")
        plt.title("Ground Truth Mask")
        plt.axis("off")

        # Predicted mask ONLY
        info = infos[i]
        plt.subplot(5, 3, 3*i + 3)
        plt.imshow(pred_masks[i], cmap="gray")
        plt.title(
            f"Predicted Mask\n"
            f"Severity: {info['severity']} | "
            f"Score: {info['severity_score']}\n"
            f"Action: {info['recommended_action']}",
            color="red"
        )
        plt.axis("off")

    plt.tight_layout()
    plt.show()

    # ---------------- Print image names separately ----------------
    print("\n===== IMAGE NAMES (DISPLAY ORDER) =====")
    for idx, name in enumerate(image_names, 1):
        print(f"{idx}. {name}")
```
