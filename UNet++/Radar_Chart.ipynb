```python
import numpy as np
import matplotlib.pyplot as plt

# Metrics
categories = ['Dice', 'IoU', 'Accuracy', 'Precision', 'Recall']
N = len(categories)

# Data
data = {
    'Rail Wear': [0.8439, 0.7409, 0.9805, 0.8688, 0.8351],
    'Rail End Batter': [0.7168, 0.5818, 0.9897, 0.7385, 0.7186],
    'Hogging of Rails': [0.7862, 0.6484, 0.9774, 0.7523, 0.8389],
    'Scabbing of Rails': [0.8439, 0.7409, 0.9805, 0.8688, 0.8351],
    'Wheel Burns': [0.7729, 0.6366, 0.9748, 0.9097, 0.6835],
    'Shelling & Black Spots': [0.8978, 0.8180, 0.9560, 0.8904, 0.9088],
    'Corrugation': [0.8045, 0.6774, 0.9653, 0.7902, 0.8295]
}

# Radar angles
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
angles += angles[:1]

# Figure
fig, ax = plt.subplots(figsize=(10,10), subplot_kw=dict(polar=True))

# Plot each defect
for defect, values in data.items():
    values += values[:1]
    ax.plot(angles, values, linewidth=2, label=defect)
    ax.fill(angles, values, alpha=0.08)

# Labels
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=12)

# Limits
ax.set_ylim(0.5, 1.0)

# Legend
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

# Title
plt.title('Multi-Metric Segmentation Performance Comparison', size=15, pad=20)

plt.show()
```
